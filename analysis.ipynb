```python
import json
moves = json.load(open('connect-four/scripts/movestring_scores.json'))
```

```python
for i, e in enumerate(moves[:-1]):
    e['moved'] = moves[i+1]['ms'][-1]
```

```python
moves = moves[:-1]
```

```python
my_moves = [e for i, e in enumerate(moves) if i % 2 == 0]
```

```python
for e in my_moves:
    e['score'] = e['scores'][int(e['moved'])]
    e['is_best'] = max(e['scores']) == e['score']
```

```python
for e in my_moves:
    if not e['is_best']:
        print(e)
```
